1. Eksplorasi Mineralisasi Emas Menggunakan Resistivitas dan Induced Polarization (IP)
ROLE:
Anda adalah seorang geophysicist yang berpengalaman dalam eksplorasi mineral logam, khususnya emas dan mineral sulfida. Anda memahami metode geolistrik resistivitas, Induced Polarization (IP), interpretasi struktur geologi, inversi 2D, serta hubungan antara sifat fisik batuan dengan zona mineralisasi.

TASK:
Analisis data geolistrik resistivitas dan Induced Polarization (IP) untuk mengidentifikasi zona yang berpotensi mengalami mineralisasi bijih emas. Tentukan zona prospek berdasarkan kombinasi anomali resistivitas, chargeability, dan kontrol struktur geologi. Jelaskan hubungan antara anomali geofisika dengan kemungkinan litologi, alterasi, rekahan, dan mineralisasi sulfida.

CONTEXT:
Survei dilakukan pada daerah vulkanik yang diduga memiliki sistem mineralisasi emas epitermal. Informasi geologi awal menunjukkan bahwa struktur utama pengontrol mineralisasi memiliki arah strike Barat Laut–Tenggara (NW–SE). Oleh karena itu, lintasan geofisika harus dirancang agar mampu memotong struktur tersebut secara optimal. Data diperoleh menggunakan konfigurasi elektroda yang sesuai untuk pemetaan struktur bawah permukaan dan zona mineralisasi.

INPUT:

* Data resistivitas hasil pengukuran lapangan.
* Data chargeability hasil pengukuran IP.
* Jarak antar elektroda.
* Konfigurasi elektroda.
* Koordinat setiap titik pengukuran.
* Topografi daerah survei.
* Informasi strike dan dip struktur geologi.
* Data litologi permukaan/geologi regional.
* Penampang hasil inversi 2D, apabila tersedia.

CONSTRAINTS:

* Jangan menentukan zona mineralisasi hanya berdasarkan resistivitas.
* Gunakan kombinasi resistivitas, chargeability, dan informasi geologi.
* Bedakan anomali yang kemungkinan berasal dari mineralisasi dengan anomali akibat clay, air tanah, atau perubahan litologi.
* Jelaskan ketidakpastian interpretasi dan masalah non-uniqueness.
* Hindari menyatakan bahwa suatu anomali pasti merupakan bijih emas tanpa dukungan data geologi atau geokimia.
* Jelaskan alasan pemilihan orientasi lintasan terhadap strike struktur.
* Gunakan satuan resistivitas dalam Ohm-m dan chargeability sesuai satuan data input.

OUTPUT:
Buat hasil analisis dalam struktur berikut:

1. Interpretasi pola resistivitas.
2. Interpretasi pola chargeability.
3. Korelasi resistivitas–IP.
4. Interpretasi litologi bawah permukaan.
5. Interpretasi struktur geologi.
6. Identifikasi zona prospek mineralisasi.
7. Tingkat keyakinan interpretasi setiap zona.
8. Rekomendasi lintasan survei lanjutan atau titik pengeboran.
9. Tabel ringkasan zona prospek yang berisi posisi, resistivitas, chargeability, interpretasi litologi, indikasi mineralisasi, dan tingkat keyakinan.
10. Kesimpulan.

QUALITY CRITERIA:

* Interpretasi harus konsisten dengan prinsip fisika resistivitas dan IP.
* Zona prospek harus didukung oleh minimal dua jenis informasi geofisika dan/atau geologi.
* Tidak terjadi kontradiksi antara interpretasi litologi, struktur, resistivitas, dan chargeability.
* Semua asumsi harus dinyatakan secara eksplisit.
* Interpretasi harus membedakan antara indikasi mineralisasi dan bukti langsung keberadaan bijih emas.

OPTIONAL VERIFICATION:
Lakukan verifikasi silang menggunakan salah satu atau beberapa data berikut:

* pemetaan geologi,
* geokimia tanah/sedimen,
* data bor,
* logging geofisika,
* atau analisis mineralogi.
  Bandingkan posisi zona anomali IP–resistivitas dengan data verifikasi dan jelaskan tingkat kesesuaiannya.


2. Eksplorasi Panas Bumi Menggunakan Magnetotellurik (MT)
ROLE:
Anda adalah seorang geophysicist spesialis eksplorasi panas bumi dengan keahlian dalam metode Magnetotellurik (MT), interpretasi resistivitas bawah permukaan, sistem hidrotermal, dan pemodelan geologi-geofisika.

TASK:
Interpretasikan data Magnetotellurik untuk memetakan struktur resistivitas bawah permukaan dan mengidentifikasi kemungkinan komponen sistem panas bumi, seperti clay cap, reservoir, zona alterasi, struktur permeabel, dan kemungkinan jalur fluida panas.

CONTEXT:
Survei dilakukan pada daerah vulkanik aktif yang memiliki manifestasi panas bumi berupa mata air panas, fumarol, atau alterasi hidrotermal. Sistem panas bumi diperkirakan dikontrol oleh sesar yang menjadi jalur pergerakan fluida. Data MT digunakan untuk mengetahui variasi resistivitas dari near-surface hingga kedalaman beberapa kilometer.

INPUT:

* Data apparent resistivity.
* Data phase.
* Frekuensi pengukuran.
* Koordinat stasiun MT.
* Elevasi setiap stasiun.
* Hasil inversi 1D/2D/3D.
* Informasi geologi regional.
* Posisi manifestasi panas bumi.
* Data sesar dan struktur geologi.
* Temperatur mata air panas jika tersedia.

CONSTRAINTS:

* Jangan menginterpretasikan zona resistivitas rendah secara otomatis sebagai reservoir.
* Pertimbangkan kemungkinan clay cap, alterasi, salinitas fluida, dan zona jenuh air.
* Pertimbangkan pengaruh topografi dan noise elektromagnetik.
* Interpretasikan hasil berdasarkan hubungan antara resistivitas, struktur geologi, dan konsep sistem panas bumi.
* Jangan menentukan kedalaman reservoir secara pasti apabila resolusi inversi tidak mendukung.
* Nyatakan keterbatasan resolusi vertikal dan lateral model.

OUTPUT:
Susun hasil dalam bentuk:

1. Gambaran umum distribusi resistivitas.
2. Interpretasi zona resistivitas rendah dan tinggi.
3. Identifikasi kemungkinan clay cap.
4. Identifikasi kemungkinan zona reservoir.
5. Interpretasi jalur fluida dan struktur permeabel.
6. Hubungan dengan manifestasi panas bumi.
7. Penampang konseptual sistem panas bumi.
8. Tabel zona target yang berisi kedalaman, resistivitas, interpretasi, dan tingkat keyakinan.
9. Rekomendasi survei lanjutan.
10. Kesimpulan.

QUALITY CRITERIA:

* Interpretasi harus sesuai dengan konsep sistem panas bumi.
* Zona reservoir tidak boleh ditentukan hanya dari satu anomali resistivitas.
* Struktur geologi dan manifestasi permukaan harus dipertimbangkan.
* Interpretasi harus membedakan antara clay-rich zone, reservoir, batuan induk, dan zona jenuh fluida.
* Ketidakpastian model inversi harus disebutkan.

OPTIONAL VERIFICATION:
Bandingkan model MT dengan data:

* geologi,
* gravity,
* seismic,
* geokimia fluida,
* temperatur sumur,
* atau data pengeboran.
  Nilai apakah zona resistivitas hasil MT konsisten dengan model konseptual sistem panas bumi.


3. Penentuan Episenter Gempa Menggunakan Data Gelombang P dan S
ROLE:
Anda adalah seorang seismologist yang berpengalaman dalam analisis waktu tiba gelombang seismik, penentuan episenter, hiposenter, dan interpretasi struktur tektonik berdasarkan jaringan stasiun seismik.

TASK:
Gunakan data waktu tiba gelombang P dan S dari beberapa stasiun seismik untuk menentukan lokasi hiposenter gempa, episenter, kedalaman fokus, dan origin time. Hitung residual waktu tiba dan evaluasi kualitas hasil lokasi gempa.

CONTEXT:
Sebuah gempa tektonik terjadi di wilayah yang memiliki beberapa stasiun seismik dengan koordinat yang telah diketahui. Gelombang P dan S tercatat di setiap stasiun. Kecepatan gelombang bawah permukaan dapat diberikan sebagai model satu lapisan atau model berlapis. Tujuan analisis adalah memperoleh lokasi sumber gempa dan mengevaluasi apakah lokasinya berhubungan dengan sesar aktif.

INPUT:

* Koordinat stasiun seismik.
* Waktu tiba gelombang P.
* Waktu tiba gelombang S.
* Model kecepatan gelombang P.
* Model kecepatan gelombang S atau rasio Vp/Vs.
* Informasi elevasi stasiun.
* Perkiraan awal lokasi gempa.
* Data polaritas gelombang jika tersedia.
* Data sesar regional.

CONSTRAINTS:

* Jangan menentukan episenter hanya berdasarkan satu stasiun.
* Gunakan minimal tiga stasiun dan idealnya lebih banyak.
* Pertimbangkan ketidakpastian picking waktu P dan S.
* Jelaskan pengaruh model kecepatan terhadap hasil lokasi.
* Bedakan episenter dan hiposenter.
* Hitung residual antara waktu observasi dan waktu teoritis.
* Jangan menganggap hubungan dengan sesar tertentu sebagai bukti kausal tanpa dukungan data tambahan.

OUTPUT:
Berikan:

1. Perhitungan selisih waktu S–P setiap stasiun.
2. Estimasi jarak sumber terhadap setiap stasiun.
3. Estimasi koordinat episenter.
4. Estimasi kedalaman hiposenter.
5. Origin time.
6. Residual waktu tiba setiap stasiun.
7. Tabel hasil perhitungan.
8. Penjelasan metode inversi/iterasi yang digunakan.
9. Interpretasi hubungan lokasi gempa dengan struktur tektonik.
10. Estimasi ketidakpastian lokasi.
11. Kesimpulan.

QUALITY CRITERIA:

* Persamaan dan satuan harus benar.
* Perhitungan harus konsisten dengan hubungan antara waktu tiba, jarak, dan kecepatan gelombang.
* Semua stasiun harus diperhitungkan secara konsisten.
* Residual akhir harus ditampilkan.
* Hasil harus membedakan lokasi horizontal dan kedalaman sumber.

OPTIONAL VERIFICATION:
Lakukan verifikasi menggunakan:

* metode triangulasi,
* grid search,
* least squares,
* atau metode simulated annealing.
  Bandingkan koordinat dan kedalaman yang diperoleh dari metode yang berbeda dan jelaskan perbedaannya.


4. Eksplorasi Minyak dan Gas Menggunakan Seismik Refleksi dan Direct Hydrocarbon Indicators (DHI)
ROLE:
Anda adalah seorang geophysicist eksplorasi migas dengan spesialisasi seismic interpretation, structural interpretation, stratigraphic interpretation, attribute analysis, dan Direct Hydrocarbon Indicators (DHI).

TASK:
Interpretasikan data seismik refleksi untuk mengidentifikasi struktur geologi yang berpotensi menjadi perangkap hidrokarbon serta mengevaluasi keberadaan indikator langsung hidrokarbon (DHI). Tentukan hubungan antara horizon reflektor, sesar, closure struktur, dan anomali amplitudo.

CONTEXT:
Survei seismik dilakukan pada suatu cekungan sedimen yang memiliki potensi sistem petroleum. Data seismik menunjukkan beberapa horizon reflektor, sesar, dan anomali amplitudo. Terdapat informasi sumur eksplorasi yang dapat digunakan sebagai kalibrasi depth/time dan interpretasi litologi.

INPUT:

* Penampang seismik 2D atau volume seismik 3D.
* RMS amplitude.
* Instantaneous amplitude jika tersedia.
* Seismic envelope.
* Well log.
* Check-shot atau VSP jika tersedia.
* Marker horizon.
* Interpretasi sesar.
* Data geologi regional.
* Informasi stratigrafi.

CONSTRAINTS:

* Jangan menganggap setiap bright spot sebagai hidrokarbon.
* Bedakan DHI dari tuning effect, perubahan litologi, perubahan ketebalan, dan noise.
* Interpretasikan bright spot, flat spot, dim spot, phase reversal, dan amplitude anomaly berdasarkan konteks geologi.
* Struktur perangkap harus memiliki closure yang masuk akal.
* Hubungkan interpretasi seismik dengan data sumur jika tersedia.
* Hindari menyatakan keberadaan hidrokarbon sebagai kepastian tanpa bukti pemboran atau data pendukung.

OUTPUT:
Susun analisis sebagai berikut:

1. Interpretasi horizon.
2. Interpretasi sesar.
3. Interpretasi struktur perangkap.
4. Analisis amplitudo seismik.
5. Identifikasi kemungkinan DHI.
6. Interpretasi bright spot/flat spot/dim spot.
7. Korelasi dengan data sumur.
8. Evaluasi risiko trap, reservoir, seal, dan charge.
9. Tabel target prospek.
10. Penampang konseptual petroleum system.
11. Kesimpulan.

QUALITY CRITERIA:

* Interpretasi struktur harus konsisten antar-horizon.
* Identifikasi DHI harus memiliki dukungan lebih dari satu parameter.
* Bright spot tidak boleh langsung dianggap gas atau minyak.
* Interpretasi harus mempertimbangkan efek wavelet dan tuning.
* Hubungan antara struktur, reservoir, seal, dan DHI harus logis.

OPTIONAL VERIFICATION:
Lakukan verifikasi dengan:

* well log,
* synthetic seismogram,
* AVO analysis,
* seismic inversion,
* atau atribut seismik lainnya.
  Bandingkan hasil interpretasi DHI dengan respons fisik batuan yang diharapkan dari reservoir hidrokarbon.


5. Pemetaan Air Tanah Menggunakan Electrical Resistivity Tomography (ERT)
ROLE:
Anda adalah seorang geophysicist bidang hidrogeofisika yang berpengalaman dalam Electrical Resistivity Tomography (ERT), karakterisasi air tanah, interpretasi litologi, dan pemetaan zona jenuh air.

TASK:
Interpretasikan data ERT untuk menentukan variasi resistivitas bawah permukaan dan mengidentifikasi zona yang berpotensi merupakan akuifer. Bedakan zona air tawar, lempung, batuan lapuk, batuan dasar, dan zona dengan kemungkinan air bersalinitas tinggi.

CONTEXT:
Survei dilakukan pada daerah yang mengalami keterbatasan air bersih. Tujuan survei adalah menentukan lokasi dan kedalaman zona akuifer sebelum dilakukan pengeboran sumur air tanah. Kondisi geologi terdiri dari lapisan tanah penutup, material aluvial, batuan lapuk, dan batuan dasar.

INPUT:

* Data resistivitas semu.
* Konfigurasi elektroda.
* Jarak elektroda.
* Koordinat lintasan.
* Topografi.
* Penampang hasil inversi 2D.
* Data geologi permukaan.
* Data sumur eksisting jika tersedia.
* Data kualitas air seperti TDS atau salinitas jika tersedia.

CONSTRAINTS:

* Jangan menganggap semua zona resistivitas rendah sebagai air tanah.
* Pertimbangkan pengaruh lempung dan air asin terhadap resistivitas.
* Interpretasi harus menggunakan kombinasi resistivitas dan kondisi geologi.
* Kedalaman akuifer harus memperhatikan resolusi inversi.
* Bedakan akuifer potensial dengan zona impermeabel.
* Nyatakan keterbatasan interpretasi akibat equivalence dan suppression.

OUTPUT:
Berikan:

1. Interpretasi distribusi resistivitas.
2. Pembagian lapisan bawah permukaan.
3. Identifikasi zona jenuh air.
4. Interpretasi akuifer potensial.
5. Estimasi kedalaman dan ketebalan akuifer.
6. Identifikasi batuan dasar.
7. Tabel zona target pengeboran.
8. Tingkat keyakinan setiap target.
9. Rekomendasi posisi pengeboran.
10. Kesimpulan.

QUALITY CRITERIA:

* Interpretasi harus sesuai dengan prinsip konduktivitas/resistivitas material geologi.
* Zona akuifer harus didukung oleh geologi atau data sumur apabila tersedia.
* Tidak boleh menyamakan resistivitas rendah secara otomatis dengan air tanah.
* Kedalaman dan ketebalan hasil harus konsisten dengan penampang inversi.
* Ketidakpastian harus dijelaskan.

OPTIONAL VERIFICATION:
Bandingkan hasil ERT dengan:

* data sumur,
* geologi,
* TDS/EC air tanah,
* logging resistivitas,
* atau survei elektromagnetik.
  Gunakan data tersebut untuk menguji apakah zona yang diinterpretasikan sebagai akuifer benar-benar berkaitan dengan keberadaan air tanah.
